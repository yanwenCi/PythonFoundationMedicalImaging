```python
import numpy as np

# Create an array with 3 rows and 4 columns.
A = np.arange(12).reshape(3, 4)
print("Original array:\n", A)
print("Shape:", A.shape)                 # (3, 4)

# Reshape: same 12 values, different shape.
reshaped = A.reshape(2, 6)
print("\nReshaped array:\n", reshaped)
print("Shape:", reshaped.shape)         # (2, 6)

# Ravel: put all values into one dimension.
flat = A.ravel()
print("\nFlat array:", flat)
print("Shape:", flat.shape)             # (12,)

# Check whether these arrays share memory with A.
print("\nReshape shares memory:", np.shares_memory(A, reshaped))
print("Ravel shares memory:", np.shares_memory(A, flat))

# Create another array with the same shape.
B = np.full((3, 4), 10)

# Vertical stack: join rows.
vertical = np.vstack([A, B])
print("\nVertical stack:\n", vertical)
print("Shape:", vertical.shape)         # (6, 4)

# Horizontal stack: join columns.
horizontal = np.hstack([A, B])
print("\nHorizontal stack:\n", horizontal)
print("Shape:", horizontal.shape)       # (3, 8)

# Mean down the rows: one mean for each column.
column_means = A.mean(axis=0)
print("\nColumn means:", column_means)  # [4. 5. 6. 7.]
print("Shape:", column_means.shape)     # (4,)

# Sum across columns: one sum for each row.
row_sums = A.sum(axis=1)
print("\nRow sums:", row_sums)           # [6 22 38]
print("Shape:", row_sums.shape)         # (3,)

# Maximum across the whole array: one value.
maximum = A.max(axis=None)
print("\nMaximum:", maximum)            # 11
print("Shape:", maximum.shape)          # ()
```

```python
import numpy as np

# Four samples and five sensors.
X = np.array([
    [20, -10,  40,  50, 120],
    [30,  40, 150,  60,  70],
    [40,  50,  60,  -5,  80],
    [50,  60,  70,  80,  90]
], dtype=float)

print("Original data:\n", X)

# 1) Find impossible readings.
mask = (X < 0) | (X > 100)
print("\nMask:\n", mask)

# 2) Replace them with NaN.
X[mask] = np.nan
print("\nCleaned data:\n", X)

# 3) Calculate one mean and standard deviation per sensor.
mu = np.nanmean(X, axis=0)
sd = np.nanstd(X, axis=0)

print("\nSensor means:", mu)
print("Sensor standard deviations:", sd)

# 4) Normalise each sensor.
Z = (X - mu) / sd
print("\nNormalised data:\n", Z)
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)

# Create data from a quadratic curve, then add random noise.
x = np.linspace(-3, 3, 20)
y = 2 * x**2 + 3 * x + 1 + rng.normal(0, 2, size=x.size)

# Use more points to draw smooth fitted curves.
x_plot = np.linspace(-3, 3, 200)

plt.scatter(x, y, color="black", label="Observed data")
plt.show()

plt.scatter(x, y, color="black", label="Observed data")
for degree in [1, 2, 3]:
    # Find the polynomial coefficients.
    coefficients = np.polyfit(x, y, deg=degree)

    # Calculate the fitted values.
    y_plot = np.polyval(coefficients, x_plot)

    plt.plot(x_plot, y_plot, label=f"Degree {degree}")
    print(f"Degree {degree} coefficients:", coefficients)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Fitting polynomials with NumPy")
plt.legend()
plt.show()
```
