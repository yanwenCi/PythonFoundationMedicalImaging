```python
class MyClass:
    variable = 'object_property'
    
    def function(self): # self allows access to all properties and methods of object          
        print(f'printed from an object method, this object has property {self.variable}') #self.variable fetches the variable property

myobjectx = MyClass()

myobjectx.variable # fetches the variable directly

myobjectx.function() # calls the function method of the object
```

```python
class MyClass:
    variable = 'object_property'

    def __init__(self, number):
        self.number = number

    def function(self):
        print(f'printed from an object method, this object has property {self.variable} and {self.number}')

myobjectx = MyClass(16)

print(myobjectx.number)

myobjectx.function()
```

```python
class MyClass:
    def __init__(self, number):
        self.number = number

    def function(self):
        print(f'printed from an object method, this object has property {self.number}')


class MyChildClass(MyClass):
    def __init__(self, number, number2):
        MyClass.__init__(self,number)
        self.number2 = number2

myobject = MyChildClass(72,10)
print(myobject.number2)
print(myobject.number)
```

```python
import time
import numpy as np
start=time.perf_counter()
A = np.arange(12).reshape(3, 4)
A[A % 2 == 0] *= -1
for row in range(A.shape[0]):
    for col in range(A.shape[1]):
        if A[row, col] % 2 == 0:
            A[row, col] *= -1
end = time.perf_counter()
print(f"time elapsed: {end-start}")
```
